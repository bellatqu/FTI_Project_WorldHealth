```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
url = "https://raw.githubusercontent.com/bellatqu/FTI_Project_WorldHealth/refs/heads/main/chronic_disease_data.csv"
df = pd.read_csv(url)

# Clean column names
df.columns = df.columns.str.strip()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,6))

top_states = df["state_name"].value_counts().head(15)

top_states.plot(kind="bar", color="steelblue")

plt.title("Top 15 States with the Most Chronic Disease Records")
plt.xlabel("State")
plt.ylabel("Number of Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#question2 answer
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
url = "https://raw.githubusercontent.com/bellatqu/FTI_Project_WorldHealth/refs/heads/main/chronic_disease_data.csv"
df = pd.read_csv(url)

# Remove rows with missing numeric values
df = df.dropna(subset=["numeric_value"])

# Calculate the average numeric value for each year
year_avg = df.groupby("year")["numeric_value"].mean()

# Create the line graph
plt.figure(figsize=(10,6))
plt.plot(year_avg.index, year_avg.values, marker="o", linewidth=2)

plt.title("Average Chronic Disease Measurement by Year")
plt.xlabel("Year")
plt.ylabel("Average Numeric Value")
plt.grid(True)

plt.show()
```

```python
# Number of records by year
print(df["year"].value_counts().sort_index())

# Average numeric value by year
print(df.groupby("year")["numeric_value"].mean())

# Count of numeric values by year
print(df.groupby("year")["numeric_value"].count())
```

```python
# to answer my question number 2 how have chronic disease measurements changed over the years?
```

```python
import matplotlib.pyplot as plt

# Calculate the average numeric value for each health topic
topic_avg = (
    df.groupby("health_topic")["numeric_value"]
      .mean()
      .sort_values(ascending=False)
)

# Create the bar chart
plt.figure(figsize=(12,6))
plt.bar(topic_avg.index, topic_avg.values)

plt.title("Average Chronic Disease Measurement by Health Topic")
plt.xlabel("Health Topic")
plt.ylabel("Average Numeric Value")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.boxplot(df["numeric_value"].dropna())

plt.title("Distribution of Chronic Disease Measurements")
plt.ylabel("Numeric Value")

plt.show()
```

```python
topic_avg = (
    df.groupby("health_topic")["numeric_value"]
      .mean()
      .sort_values()
)

plt.figure(figsize=(10,6))
plt.barh(topic_avg.index, topic_avg.values)

plt.title("Average Chronic Disease Measurement by Health Topic")
plt.xlabel("Average Numeric Value")
plt.ylabel("Health Topic")

plt.show()
```

```python
demo_avg = (
    df.groupby("demographic_group")["numeric_value"]
      .mean()
      .sort_values()
)

plt.figure(figsize=(10,6))
plt.barh(demo_avg.index, demo_avg.values)

plt.title("Average Chronic Disease Measurement by Demographic Group")
plt.xlabel("Average Numeric Value")
plt.ylabel("Demographic Group")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Top 10 health topics by average numeric value
topic_avg = (
    df.groupby("health_topic")["numeric_value"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))
plt.bar(topic_avg.index, topic_avg.values)

plt.title("Top 10 Chronic Disease Categories by Average Measurement")
plt.xlabel("Health Topic")
plt.ylabel("Average Numeric Value")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```

```python
indicator_avg = (
    df.groupby("indicator_metric")["numeric_value"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(12,6))
plt.barh(indicator_avg.index, indicator_avg.values)

plt.title("Top 10 Indicator Metrics by Average Measurement")
plt.xlabel("Average Numeric Value")
plt.tight_layout()

plt.show()
```

```python
topic_count = df["health_topic"].value_counts()

plt.figure(figsize=(10,6))
plt.bar(topic_count.index, topic_count.values)

plt.title("Number of Records by Health Topic")
plt.xlabel("Health Topic")
plt.ylabel("Number of Records")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()

plt.show()
```
